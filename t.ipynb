```python
from glob import glob
import json

files = glob('results/**/*.json', recursive=True)

sub = []
for f in files:
    if 'log' not in f:
        with open(f, 'r', encoding='utf-8') as fp:
            sub.append(json.load(fp))
```

```python
print(sub[0])
```

```python
wrong = 0
correct  = 0
wrong_fix = 0
wrong_gen = 0
both_fail = 0

for s in sub:
    if not s['anchor']['result']:
        wrong+=1
    elif s['llm_fix_code']['result'] and s['llm_generate_code']['result']:
        correct+=1
    elif not s['llm_fix_code']['result'] and s['llm_generate_code']['result']:
        wrong_fix+=1
    elif s['llm_fix_code']['result'] and not s['llm_generate_code']['result']:
        wrong_gen+=1
    else:
        both_fail+=1
    
print(f'Wrong: {wrong} / {len(sub)}')
print(f'Correct: {correct} / {len(sub)}')
print(f'Wrong fix: {wrong_fix} / {len(sub)}')
print(f'Wrong gen: {wrong_gen} / {len(sub)}')
print(f'Both fail: {both_fail} / {len(sub)}')
```
